```python
import pandas as pd

def find_s(data):
    # Get all attributes except the target column
    attributes = data.columns[:-1]
    target = data.columns[-1]

    # Initialize hypothesis with most specific values
    hypothesis = ['0'] * len(attributes)

    # Process each training example
    for i, row in data.iterrows():

        # Consider only positive examples
        if row[target] == 'Yes':

            for j, attribute in enumerate(attributes):

                # First positive example
                if hypothesis[j] == '0':
                    hypothesis[j] = row[attribute]

                # If values are different, generalize to '?'
                elif hypothesis[j] != row[attribute]:
                    hypothesis[j] = '?'

    return hypothesis


# Example dataset
data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'No'],
    ['Sunny', 'Warm', 'High',   'Weak',   'Yes']
], columns=['Sky', 'Temperature', 'Humidity', 'Wind', 'EnjoySport'])


# Apply Find-S
hypothesis = find_s(data)

print("Final Hypothesis:")
print(hypothesis)
```
